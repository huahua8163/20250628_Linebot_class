```python
from google import genai
import os
from dotenv import load_dotenv

# 載入環境變數
load_dotenv(override=True)

# 使用環境變數中的 API 密鑰
client = genai.Client(api_key=os.environ['GEMINI_API_KEY'])

# 使用 stream 版本
response = client.models.generate_content_stream(
    model="gemini-1.5-flash",  # 改成正確的模型名稱
    contents=["AI是如何工作的(請使用繁體中文回答)?"],
)

for chunk in response:
    print(chunk.text, end="")
```

```python
import PIL.Image
from google import genai
from IPython.display import display, Markdown
import os
from dotenv import load_dotenv

# 載入環境變數
load_dotenv(override=True)

# 使用環境變數中的 API 密鑰
client = genai.Client(api_key=os.environ['GEMINI_API_KEY'])

# 檢查圖片檔案是否存在
if os.path.exists('bear.jpg'):
    print("找到 bear.jpg 檔案")
    image = PIL.Image.open('bear.jpg')
    
    response = client.models.generate_content(
        model="gemini-1.5-flash",
        contents=[image, "請告訴我這是什麼動物,還有關於它的一些資訊"]
    )
    
    display(Markdown(response.text))
else:
    print("找不到 bear.jpg 檔案")
    # 測試純文字功能
    response = client.models.generate_content(
        model="gemini-1.5-flash",
        contents=["你好！請用繁體中文回答：什麼是人工智慧？"]
    )
    
    display(Markdown(response.text))
```
